```python
import math
import time
import numpy as np
import torch
import matplotlib.pyplot as plt
```

```python
## define PDF of Gaussian distribution

def Gaussian(x, mu, sigma):
    p = 1 / math.sqrt(2 * math.pi * sigma**2)
    return p * np.exp(-0.5/sigma**2 * (x-mu)**2)
```

```python

# x 轴
x = np.arange(-7, 7, 0.01)

# (均值, 标准差)
params = [(0, 1), (0, 2), (3, 1)]

# 绘图
for mu, sigma in params:
    plt.plot(
        x,
        Gaussian(x, mu, sigma),
        label=f"mean {mu}, std {sigma}"
    )

plt.xlabel("x")
plt.ylabel("p(x)")
plt.legend()
plt.grid(True)

plt.show()
```
