```python
import math
import time
import numpy as np
import torch
```

```python
## 测试矢量化加速
n = 10000
a = torch.ones([n])
b = torch.ones([n])
```

```python
class Timer:
    """记录多次运行时间"""

    def __init__(self):
        self.times = []
        self.start()

    def start(self):
        """启动计时器"""
        self.tik = time.time()

    def stop(self):
        """停止计时器，并将本次运行时间记录在列表中"""
        elapsed = time.time() - self.tik
        self.times.append(elapsed)
        return self.times[-1]

    def avg(self):
        """返回平均运行时间"""
        return sum(self.times) / len(self.times)

    def sum(self):
        """返回运行时间总和"""
        return sum(self.times)

    def cumsum(self):
        """返回累计运行时间"""
        return np.array(self.times).cumsum().tolist()
```

```python
## 使用for loop的所需时间
c = torch.zeros([n])

timer = Timer()

for i in range(n):
    c[i] = a[i] + b[i]
    
f'{timer.stop():.5f} sec'
```
