```python
import torch
from torch import nn
from torch.nn import functional as F
```

```python
## save & load
x = torch.arange(4)
torch.save(x,"x-file")

x2 = torch.load("x-file")
x2
```

```python
## save & load tensor list
y = torch.zeros(4)

torch.save([x,y],"x-files")

x2,y2 = torch.load("x-files")
(x2,y2)
```

```python
## save & load dict
mydict = {
    "x":x,
    "y":y
}
torch.save(mydict,"mydict")

mydict2 = torch.load("mydict")
mydict2
```

```python
class MLP(nn.Module):
    def __init__(self):
        super().__init__()
        self.hidden = nn.Linear(20,256)
        self.out = nn.Linear(256,10)
    def forward(self,X):
        return self.out(F.relu(self.hidden(X)))
net = MLP()
X = torch.randn((2,20))
Y = net(X)

## 保存参数
torch.save(net.state_dict(),"mpl-params")

## clone一个备份来load 参数
clone = MLP()
clone.load_state_dict(torch.load("mpl-params"))
clone.eval()
```
