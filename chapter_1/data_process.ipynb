```python
import torch
import pandas as pd
import os
```

```python
## 创建数据集

os.makedirs(os.path.join('..', 'data'), exist_ok=True)

data_file = os.path.join('..', 'data', 'house_tiny.csv')

with open(data_file, 'w') as f:
    f.write('NumRooms,Alley,Price\n')  # 列名
    f.write('NA,Pave,127500\n')        # 每行表示一个数据样本
    f.write('2,NA,106000\n')
    f.write('4,NA,178100\n')
    f.write('NA,NA,140000\n')
```

```python
data = pd.read_csv('/Users/gaofan/Projects/dl-by-hand_2/data/house_tiny.csv')

print(data)
```

```python
## 处理缺失值

## 分割 input 和 output
input = data.iloc[: ,0:2]

output = data.iloc[: ,2]

## 均值填补缺失值
input.iloc[:, 0] = input.iloc[:, 0].fillna(
    input.iloc[:, 0].mean()
)

## one-hot encoding
input = pd.get_dummies(input, columns=["Alley"], dummy_na=True, dtype=int)

print(input)
```

```python
## 转换为tensor
X = torch.tensor(input.to_numpy(dtype = float))

y = torch.tensor(output.to_numpy(dtype = float))\

print(X,y)
```
