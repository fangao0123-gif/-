```python
import torch
```

```python
## 按元素运算 有一个float都变float

X = torch.tensor([1.0, 2, 4, 8])
Y = torch.tensor([2, 2, 2, 2])


print("X + Y:", X + Y)

print("X - Y:", X - Y)

print("X * Y:", X * Y) ## 按元素相乘

print("X / Y:", X / Y) ## 按元素相除

print("X ** Y:", X ** Y) ## 按元素幂运算 X的Y次幂
```

```python
## 拼接tensor
X = torch.arange(12, dtype=torch.float32).reshape((3,4))

Y = torch.tensor([[2.0, 1, 4, 3], [1, 2, 3, 4], [4, 5, 6, 7]])

## 按行拼接
print("torch.cat((X, Y), dim=0):", torch.cat((X, Y), dim=0))

## 按列拼接
print("torch.cat((X,Y), dim=1):", torch.cat((X,Y), dim=1))


## 利用逻辑运算符构筑tensor
print("X == Y:", X == Y)
```

```python
## Broadcasting机制
X = torch.arange(3).reshape((3,1))
Y = torch.arange(2).reshape((1,2))

print("X + Y:", X + Y)
```

```python
## 节约内存的原地操作
X = torch.arange(12, dtype=torch.float32).reshape((3,4))
Y = torch.tensor([[2.0, 1, 4, 3], [1, 2, 3, 4], [4, 5, 6, 7]])

before = id(Y)
Y = Y + X
print(id(Y) == before) ## False

```

```python
## 节约内存的原地操作
X = torch.arange(12, dtype=torch.float32).reshape((3,4))
Y = torch.tensor([[2.0, 1, 4, 3], [1, 2, 3, 4], [4, 5, 6, 7]])

before = id(Y)
Y += X
print(id(Y) == before) ## True

```

```python
## 节约内存的原地操作
X = torch.arange(12, dtype=torch.float32).reshape((3,4))
Y = torch.tensor([[2.0, 1, 4, 3], [1, 2, 3, 4], [4, 5, 6, 7]])

Z = torch.zeros_like(Y)
print("id(Z):", id(Z))

Z[:] = Y + X

print("id(Z):", id(Z)) ## Z的id没有变
```
