```python
import torch
import pandas as pd
import numpy as np
```

```python
## 一维tensor 就是数学中的 vector
x = torch.arange(4)

x[3]
```

```python
## matrix 
A = torch.arange(20).reshape(5,4)

print(A)

## A的转置
A.T
```

```python
## 3维tensor
X = torch.arange(24).reshape(2,4,3)
X
```

```python
## 矩阵运算
A = torch.arange(20).reshape(5,4)

B = A.clone()
## 加法
print(A + B)

## hadamard积
print(A * B)

## 降维（沿两个轴）
print(A.sum())

## 降维 （沿axis = 0）
A_sum_axis0 = A.sum(axis = 0)
print(A_sum_axis0)

## 降维 （沿axis = 1）
A_sum_axis1 = A.sum(axis = 1)
print(A_sum_axis1)
```

```python
## 非降维求和
A = torch.arange(20).reshape(5,4)

sum_A = A.sum(axis = 1, keepdims = True)
print(sum_A)

## 利用broadcasting A/sum_A
print(A/sum_A)

## 累积元素和
print(A.cumsum(axis = 0))
```

```python
## dot product
x = torch.tensor([0.0,1,2,3])
y = torch.ones(4, dtype = torch.float32)

print(torch.dot(x,y))
```

```python
## marix-vector product
A = torch.arange(20, dtype = torch.float32).reshape(5,4)
x = torch.ones(4, dtype = torch.float32)

mv_product = torch.mv(A,x)
print(mv_product)
```

```python
## matrix-matrix multipltion

A = torch.arange(20, dtype = torch.float32).reshape(5,4)
x = torch.ones(4,3, dtype = torch.float32)

mm = torch.mm(A,x)
print(mm)
```

```python
## norm

## L2 norm
x = torch.tensor([3.0, -4.0])
L2_norm = torch.norm(x)
L2_norm

## L1 norm

L1_norm = torch.abs(x).sum()
L1_norm

## L_F norm
x = torch.ones(4,9)
LF_norm = torch.norm(x)
LF_norm
```
