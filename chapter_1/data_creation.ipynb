```python
import torch
```

```python
## 创建一个row vector
x = torch.arange(12)
x
```

```python
## 查看 shape 和 size
x.shape

print("Shape of x:", x.shape)

print("Size of x:", x.size())
```

```python
## reshape 改变tensor形状
X = x.reshape(3, 4)
X

## 也可以使用 -1 来自动推断某一维的大小
Y = x.reshape(-1, 4)
Y

```

```python
## 创建 全0 tensor 
Z = torch.zeros((2,3,3))

Z
```

```python
## 创建 全1 tensor
One = torch.ones((2,3,3))
One
```

```python
## 从标准正态分布中随机采样创建tensor
Normal = torch.randn((1,3,3))
Normal
```
