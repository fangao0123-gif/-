```python
import torch
from torch import nn
from d2l import torch as d2l
```

```python
## define network
net = nn.Sequential(
    nn.Flatten(),
    nn.Linear(784,256),
    nn.ReLU(),
    nn.Linear(256,10)
)
```

```python
## initialization

def init_weights(m):
    ## 使用 isinstance 判断层类型
    if isinstance(m, nn.Linear):
      nn.init.normal_(m.weight, std=0.01)
        

net.apply(init_weights)
```
