```python
import torch
from torch import nn
import matplotlib.pyplot as plt
```

```python
## plot relu(x)
x = torch.arange(-8.0,8.0,0.1,requires_grad=True)
y = torch.relu(x)

plt.plot(x.detach(), y.detach(), label='relu(x)')
plt.xlabel('x')
plt.ylabel('relu(x)')
plt.legend()
plt.show()
```

```python
## plot grad of relu(x)
y.sum().backward(retain_graph=True)
plt.plot(x.detach(),x.grad,label = "grad of relu(x)")
plt.xlabel('x')
plt.ylabel('grad of relu(x)')
plt.legend()
plt.show()
```

```python
## plot sigmoid(x)
y = torch.sigmoid(x)
plt.plot(x.detach(),y.detach(),label='sigmoid(x)')
plt.xlabel('x')
plt.ylabel('sigmoid(x)')
plt.show
```

```python
## grad of sigmoid(x)
x.grad.data.zero_()
y.sum().backward(retain_graph=True)
plt.plot(x.detach(),x.grad,label='grad of sigmoid(x)')
plt.xlabel('x')
plt.ylabel('grad of sigmoid(x)')
plt.legend()
plt.show()
```
